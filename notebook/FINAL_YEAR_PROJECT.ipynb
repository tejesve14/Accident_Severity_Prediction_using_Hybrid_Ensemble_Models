FEATURE ENGINEERING

```python
# convert the 'Date' column to datetime format
df['Time']= pd.to_datetime(df['Time'])
```

```python
# Convert Time to datetime and extract hour
df['Hour'] = pd.to_datetime(df['Time'], errors='coerce').dt.hour

# Create Time_Band
def time_band(hour):
    if 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 18:
        return "Afternoon"
    elif 18 <= hour < 24:
        return "Evening"
    else:
        return "Night"

df['Time_Band'] = df['Hour'].apply(time_band)
```

```python
# Drop 'Time' and 'Hour' and save the result back to df
df = df.drop(columns=['Time', 'Hour'])
```

SPLITTING TEST AND MODEL TRAINING


```python
X = df.drop("Accident_severity", axis=1)
y = df["Accident_severity"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nTrain distribution:")
print(y_train.value_counts())

print("\nTest distribution:")
print(y_test.value_counts())
```

```python
from sklearn.metrics import accuracy_score, classification_report

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
```

LOGISTIC REGRESSION

```python
log_model = LogisticRegression(
    max_iter=5000,
    class_weight="balanced"
)

log_model.fit(X_train, y_train)

y_pred_log = log_model.predict(X_test)

print("Logistic Regression")
print("Accuracy:", accuracy_score(y_test, y_pred_log))
print(classification_report(y_test, y_pred_log))
```

DECISION TREE

```python
dt_model = DecisionTreeClassifier(
    max_depth=15,
    class_weight="balanced",
    random_state=42
)

dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)

print("Decision Tree")
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt))
```

RANDOM FOREST

```python
rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=20,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf))
```

GRADIENT BOOSTING

```python
gb_model = GradientBoostingClassifier(
    n_estimators=250,
    learning_rate=0.08,
    max_depth=5
)

gb_model.fit(X_train, y_train)

y_pred_gb = gb_model.predict(X_test)

print("Gradient Boosting")
print("Accuracy:", accuracy_score(y_test, y_pred_gb))
print(classification_report(y_test, y_pred_gb))
```

XG BOOST

```python
xgb_model = XGBClassifier(
    n_estimators=400,
    max_depth=8,
    learning_rate=0.05,
    subsample=0.9,
    colsample_bytree=0.9,
    eval_metric="mlogloss",
    random_state=42
)

xgb_model.fit(X_train, y_train)

y_pred_xgb = xgb_model.predict(X_test)

print("XGBoost")
print("Accuracy:", accuracy_score(y_test, y_pred_xgb))
print(classification_report(y_test, y_pred_xgb))
```

HYBRID STACKING ENSEMBLE MODEL

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.ensemble import StackingClassifier, RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. Define your original Top 3 Base Models
rf_model = RandomForestClassifier(n_estimators=400, max_depth=20, random_state=42)
xgb_model = XGBClassifier(n_estimators=400, learning_rate=0.05, max_depth=6, random_state=42)
gb_model = GradientBoostingClassifier(n_estimators=300, learning_rate=0.05, random_state=42)

base_learners = [
    ('rf', rf_model),
    ('xgb', xgb_model),
    ('gb', gb_model)
]

# 2. Define the Meta-Learner
meta_model = LogisticRegression(max_iter=2000)

# 3. Create the Stacking Classifier
stack_model = StackingClassifier(
    estimators=base_learners,
    final_estimator=meta_model,
    stack_method='predict_proba',
    cv=5,
    n_jobs=-1
)


smote_stack_pipeline = ImbPipeline([
    ('smote', SMOTE(random_state=42)),
    ('stacking', stack_model)
])

# 5. Train the Pipeline
print("Training Hybrid Stacking... ")
smote_stack_pipeline.fit(X_train, y_train)

# 6. Predict and Print Results
y_pred_smote = smote_stack_pipeline.predict(X_test)

print(f"Accuracy: {accuracy_score(y_test, y_pred_smote)}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_smote))
```

```python
import streamlit as st
import pandas as pd
import pickle
import shap
import matplotlib.pyplot as plt

# ---------------- LOAD MODELS ----------------
stack_model = pickle.load(open("stack_model.pkl", "rb"))
rf_model = pickle.load(open("rf_model.pkl", "rb"))

st.set_page_config(page_title="Accident Predictor", layout="wide")

st.title("🚗 Accident Severity Prediction")

# ---------------- MAPPINGS (IMPORTANT) ----------------
age_map = {"18-30": 1, "31-50": 2, "Over 51": 3}
experience_map = {"1-2yr": 0, "2-5yr": 1, "5-10yr": 2, "Above 10yr": 3}
vehicle_map = {"Automobile": 0, "Motorcycle": 1, "Lorry": 2}
area_map = {"Urban": 0, "Rural": 1}
lanes_map = {"One way": 0, "Two way": 1}
road_map = {"Straight": 0, "Curve": 1}
junction_map = {"No junction": 0, "T-junction": 1, "Cross": 2}
surface_map = {"Dry": 0, "Wet": 1}
light_map = {"Daylight": 0, "Night": 1}
weather_map = {"Normal": 0, "Rainy": 1, "Fog": 2}
collision_map = {"Vehicle-Vehicle": 0, "Vehicle-Object": 1}
movement_map = {"Moving": 0, "Stopped": 1}
casualty_class_map = {"Driver": 0, "Passenger": 1}
pedestrian_map = {"Crossing": 0, "Walking": 1}
cause_map = {"Overspeed": 0, "Overtaking": 1, "Careless": 2}
time_map = {"Morning": 0, "Afternoon": 1, "Evening": 2, "Night": 3}

# ---------------- INPUT UI ----------------
col1, col2 = st.columns(2)

with col1:
    age = st.selectbox("Age Band", list(age_map.keys()))
    experience = st.selectbox("Driving Experience", list(experience_map.keys()))
    vehicle = st.selectbox("Vehicle Type", list(vehicle_map.keys()))
    area = st.selectbox("Area", list(area_map.keys()))
    lanes = st.selectbox("Lanes", list(lanes_map.keys()))
    road_align = st.selectbox("Road Alignment", list(road_map.keys()))
    junction = st.selectbox("Junction", list(junction_map.keys()))
    surface = st.selectbox("Road Surface", list(surface_map.keys()))
    light = st.selectbox("Light", list(light_map.keys()))
    weather = st.selectbox("Weather", list(weather_map.keys()))

with col2:
    collision = st.selectbox("Collision Type", list(collision_map.keys()))
    vehicles = st.slider("Vehicles Involved", 1, 10, 1)
    casualties = st.slider("Casualties", 1, 10, 1)
    movement = st.selectbox("Vehicle Movement", list(movement_map.keys()))
    casualty_class = st.selectbox("Casualty Class", list(casualty_class_map.keys()))
    pedestrian = st.selectbox("Pedestrian Movement", list(pedestrian_map.keys()))
    cause = st.selectbox("Cause", list(cause_map.keys()))
    time_band = st.selectbox("Time Band", list(time_map.keys()))

# ---------------- PREDICT ----------------
if st.button("Predict"):

    # Convert to encoded values
    input_data = pd.DataFrame([{
        "Age_band_of_driver": age_map[age],
        "Driving_experience": experience_map[experience],
        "Type_of_vehicle": vehicle_map[vehicle],
        "Area_accident_occured": area_map[area],
        "Lanes_or_Medians": lanes_map[lanes],
        "Road_allignment": road_map[road_align],
        "Types_of_Junction": junction_map[junction],
        "Road_surface_conditions": surface_map[surface],
        "Light_conditions": light_map[light],
        "Weather_conditions": weather_map[weather],
        "Type_of_collision": collision_map[collision],
        "Number_of_vehicles_involved": vehicles,
        "Number_of_casualties": casualties,
        "Vehicle_movement": movement_map[movement],
        "Casualty_class": casualty_class_map[casualty_class],
        "Pedestrian_movement": pedestrian_map[pedestrian],
        "Cause_of_accident": cause_map[cause],
        "Time_Band": time_map[time_band]
    }])

    # ---------------- PREDICTION ----------------
    prediction = stack_model.predict(input_data)[0]

    if prediction == 0:
        st.error("🚨 Fatal Injury")
    elif prediction == 1:
        st.warning("⚠️ Serious Injury")
    else:
        st.success("✅ Slight Injury")

    # ---------------- SHAP ----------------
    st.subheader("🔍 Explainability (SHAP)")

    explainer = shap.TreeExplainer(rf_model)
    shap_values = explainer.shap_values(input_data)

    if isinstance(shap_values, list):
        shap_values = shap_values[1]

    fig, ax = plt.subplots()
    shap.force_plot(
        explainer.expected_value,
        shap_values[0],
        input_data,
        matplotlib=True,
        show=False
    )

    st.pyplot(fig)
```
